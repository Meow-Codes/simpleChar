```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM, Embedding, Bidirectional
```

```python
# https://www.codecademy.com/catalog/subject/data-science
text = '''Data scientists try to make sense of the data that’s all around us.
Taking a data science course can help you make informed decisions, create beautiful visualizations,
and even try to predict future events through Machine Learning. If you’re curious about what you can
learn about the world using the data produced every day, then data science might be for you!'''
```

```python
tokens = text.split()
raw_text = ' '.join(tokens)
print(raw_text)
```

```python
chars = sorted(list(set(text)))
print(chars)
mapping = dict((c,i) for i, c in enumerate(chars))
print(mapping)
vocab = len(mapping)
print(vocab)
```

```python
length = 10
sequences = list()
for i in range(length, len(raw_text)):
  seq = raw_text[i-length:i+1]
  sequences.append(seq)
print(sequences)
```

```python
new_sequence = list()
for seq in sequences:
 encoded_seq = [mapping[char] for char in seq]
 new_sequence.append(encoded_seq)
print(new_sequence)
```

```python
sequences = np.array(new_sequence)
X, y = sequences[:,:-1], sequences[:,-1]
```

```python
X
```

```python
y
```

```python
sequences = [to_categorical(x, num_classes=vocab) for x in X]
X = np.array(sequences)
y = to_categorical(y, num_classes=vocab)
```

```python
X
```

```python
y
```

```python
def create_model(X):
  model = Sequential()
  model.add(Bidirectional(LSTM(50, return_sequences=False), input_shape=(X.shape[1], X.shape[2])))
  model.add(Dense(vocab, activation='softmax'))
  model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
  model.summary()
  return model

model = create_model(X)
model.fit(X,y,epochs=200)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical
def generate_sequence (model, mapping, max_length, text, n_chars):
  text_t = text.lower() # Convert input text to lowercase
  for _ in range(n_chars):
    encoded = [mapping[char] for char in text_t]
    encoded = pad_sequences([encoded], maxlen = max_length, truncating = 'pre')
    encoded = to_categorical(encoded, num_classes = vocab)
    yhat = model.predict(encoded)
    yhat = np.argmax(yhat, axis=-1)
    output = ''
    for char, index in mapping.items():
      if index == yhat:
        output = char
        break
    text_t += output
  return text_t
generate_sequence (model, mapping, 10, 'Data Scientists', 30)
```
