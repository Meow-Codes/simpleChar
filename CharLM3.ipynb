```python
from numpy import *
from pylab import *

# Parameters
R = 4.0        # Earth's orbit radius around Sun
r1 = 1.0       # Moon's orbit radius around Earth
r2 = 0.25      # Chandrayaan's orbit radius around Moon
r3 = 0.0625    # Satellite1's orbit radius around Chandrayaan
r4 = 0.015625  # Satellite2's orbit radius around Chandrayaan

wp = 2.0       # Earth's angular speed
wm = 30.0      # Moon's angular speed
ws = 450.0     # Chandrayaan's angular speed
w3 = 6750.0    # Satellite1's angular speed
w4 = 101250.0  # Satellite2's angular speed

t = linspace(0.0, 10.0, 100000)

# Earth
x = R*cos(wp*t)
y = R*sin(wp*t)

# Moon
x0 = x + r1*cos(wm*t)
y0 = y + r1*sin(wm*t)

# Chandrayaan
x1 = x0 + r2*cos(ws*t)
y1 = y0 + r2*sin(ws*t)

# Satellite1 around Chandrayaan
x2 = x1 + r3*cos(w3*t)
y2 = y1 + r3*sin(w3*t)

# Satellite around Chandrayaan
x3 = x2 + r4*cos(w4*t)
y3 = y2 + r4*sin(w4*t)

# Plot all trajectories
plot(x, y)
plot(x0, y0)
plot(x1, y1)
plot(x2, y2)
plot(x3, y3)
legend(["Earth's Trajectory", "Moon's Trajectory", "Chandrayaan's Trajectory", "Satellite1's Trajectory", "Satellite2's Trajectory"])
title("Trajectories w.r.t Sun")
axis('equal')
show()
```
